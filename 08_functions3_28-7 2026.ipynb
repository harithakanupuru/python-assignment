```python
def calculate_factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
    return factorial
number = int(input("Enter number: "))
result = calculate_factorial(number)
print("Factorial:", result)
```

Assignment 35: Fibonacci Function

Write a function named generate_fibonacci(). Requirements Accept the number of terms. Return the complete Fibonacci series as a list. Handle zero and negative term counts. Do not use recursion in this assignment. Example Input Enter number of terms: 7 Expected Output [0, 1, 1, 2, 3, 5, 8]



```python
def generate_fibonacci(number_of_terms):
    if number_of_terms <= 0:
        return "Invalid number of terms."
    fibonacci_series = [0, 1]
    while len(fibonacci_series) < number_of_terms:
        next_number = fibonacci_series[-1] + fibonacci_series[-2]
        fibonacci_series.append(next_number)
    return fibonacci_series
number_of_terms = int(input("Enter number of terms: "))
fibonacci_series = generate_fibonacci(number_of_terms)
print("Fibonacci Series:", fibonacci_series)
```

```python
from functools import reduce
employees = [
    {"id": 101, "name": "Ravi", "department": "IT", "salary": 60000, "performance_score": 88, "experience": 5},
    {"id": 102, "name": "Anitha", "department": "HR", "salary": 45000, "performance_score": 72, "experience": 4},
    {"id": 103, "name": "Kiran", "department": "IT", "salary": 75000, "performance_score": 94, "experience": 7},
    {"id": 104, "name": "Meena", "department": "Finance", "salary": 55000, "performance_score": 81, "experience": 6},
    {"id": 105, "name": "Arjun", "department": "IT", "salary": 40000, "performance_score": 65, "experience": 2}
]
company_name = "Edukron"
def display_employees(emp_list):
    """Prints all employee details in readable format"""
    if not emp_list:
        print("No employees found.")
        return
    for emp in emp_list:
        print(f"ID:{emp['id']} | Name:{emp['name']} | Dept:{emp['department']} | "
              f"Salary:{emp['salary']} | Score:{emp['performance_score']} | Exp:{emp['experience']}")
def get_employee_names(emp_list):
    return list(map(lambda e: e['name'], emp_list))
def get_high_performers(emp_list):
    return list(filter(lambda e: e['performance_score'] >= 80, emp_list))
def filter_by_department(emp_list, dept):
    return list(filter(lambda e: e['department'].lower() == dept.lower(), emp_list))
def calculate_revised_salaries(emp_list):
    """15% if >=90, 10% if 80-89, 5% if 70-79, else 0. Does not modify original"""
    def revise(emp):
        score = emp['performance_score']
        if score >= 90: inc = 0.15
        elif score >= 80: inc = 0.10
        elif score >= 70: inc = 0.05
        else: inc = 0
        new_emp = emp.copy()
        new_emp['revised_salary'] = round(emp['salary'] * (1 + inc), 2)
        return new_emp
    return list(map(revise, emp_list))
def total_salary_expense(emp_list, key='salary'):
    return reduce(lambda total, e: total + e.get(key, 0), emp_list, 0)
def find_highest_paid(emp_list):
    if not emp_list: return None
    return reduce(lambda a, b: a if a['salary'] > b['salary'] else b, emp_list)
def sort_employees(emp_list, key):
    valid_keys = ['name', 'salary', 'performance_score', 'experience']
    if key not in valid_keys:
        print("Invalid sort key")
        return emp_list
    return sorted(emp_list, key=lambda e: e[key])
def create_bonus_calculator(percent):
    """Outer function takes %, inner function takes salary"""
    def calculate_bonus(salary):
        return round(salary * percent / 100, 2)
    return calculate_bonus
def generate_department_report(emp_list, dept):
    dept_employees = filter_by_department(emp_list, dept)
    def num_employees(): # Local
        return len(dept_employees)
    def total_sal():
        return total_salary_expense(dept_employees)
    def avg_sal():
        return round(total_sal() / num_employees(), 2) if num_employees() > 0 else 0
    def avg_perf():
        scores = list(map(lambda e: e['performance_score'], dept_employees))
        return round(sum(scores) / len(scores), 2) if scores else 0
    report_title = f"{dept} Department Report"
    return {
        "Company": company_name,
        "Report Title": report_title,
        "Department": dept,
        "Number of Employees": num_employees(),
        "Total Salary": total_sal(),
        "Average Salary": avg_sal(),
        "Average Performance Score": avg_perf()
    }
def main():
    revised_employees = calculate_revised_salaries(employees)
    bonus5 = create_bonus_calculator(5)
    bonus10 = create_bonus_calculator(10)
    bonus15 = create_bonus_calculator(15)
    while True:
        print("\n========== Employee Performance System ==========")
        print("1. Display all employees")
        print("2. Display employee names")
        print("3. Display high performers")
        print("4. Search by department")
        print("5. Calculate revised salaries")
        print("6. Calculate total salary expense")
        print("7. Find highest-paid employee")
        print("8. Sort employees")
        print("9. Generate department report")
        print("10. Calculate Bonus")
        print("11. Exit")
        choice = input("Enter your choice: ")
        if choice == '1':
            display_employees(employees)
        elif choice == '2':
            print("Employee Names:", get_employee_names(employees))
        elif choice == '3':
            display_employees(get_high_performers(employees))
        elif choice == '4':
            dept = input("Enter department name: ")
            result = filter_by_department(employees, dept)
            display_employees(result) if result else print("Department not found")
        elif choice == '5':
            print("--- Revised Salaries ---")
            display_employees(revised_employees)
        elif choice == '6':
            current = total_salary_expense(employees)
            revised = total_salary_expense(revised_employees, 'revised_salary')
            print(f"Current Total: {current}")
            print(f"Revised Total: {revised}")
            print(f"Additional Expense: {revised - current}")
        elif choice == '7':
            print("Highest Paid Employee:")
            print(find_highest_paid(employees))
        elif choice == '8':
            key = input("Sort by name/salary/performance_score/experience: ")
            display_employees(sort_employees(employees, key))
        elif choice == '9':
            dept = input("Enter department for report: ")
            print(generate_department_report(employees, dept))
        elif choice == '10':
            sal = float(input("Enter salary: "))
            print(f"5% Bonus: {bonus5(sal)}, 10% Bonus: {bonus10(sal)}, 15% Bonus: {bonus15(sal)}")
        elif choice == '11':
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")
if __name__ == "__main__":
    main()
```
